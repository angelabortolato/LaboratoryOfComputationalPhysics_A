2\.b How many even numbers are there?

```bash

even=0

while read l; do
for n in $l; do
if [[ $n =~ ^[0-9]+$ ]]; then
if (( n % 2 == 0 )); then
(( even+=1 ))
fi
fi
done
done <data.txt

echo "$even"
```


2\.c Distinguish the entries on the basis of `sqrt(X^2 + Y^2 + Z^2)` is greater or smaller than `100*sqrt(3)/2`. Count the entries of each of the two groups 

```bash

threshold=$(echo "sqrt(3)*100/2" | bc)
great=0
small=0
while read line; do
    x=$(echo "$line" | cut -d' ' -f1)
    y=$(echo "$line" | cut -d' ' -f2)
    z=$(echo "$line" | cut -d' ' -f3)
    r=$(echo "sqrt($x*$x+$y*$y+$z*$z)" | bc)
    if (( r > threshold)); then
    (( great+=1 ))
    else (( small+=1 ))
    fi
done <data.txt
echo "greater are $great , smaller are $small"
```


2\.d Make `n` copies of data.txt (with `n` an input parameter of the script), where the i-th copy has all the numbers divided by i (with `1<=i<=n`).

```bash
 m=5
for i in $(seq 1 $m); do
 while read l; do
    for n in $l; do
        if [[ $n =~ ^[0-9]+$ ]]; then
        nn=$(echo "scale=3;$n/$i" | bc)
        echo "$nn " >> data_$i.csv
        fi
    done
 echo "newline" >> data_$i.csv
 done <data.txt
 sed -i 's/\n/ /g' data_$i.txt
 sed -i 's/newline/\n/g' data_$i.txt
done 
```
